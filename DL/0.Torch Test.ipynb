```python
import torch
import numpy as np

w = torch.tensor([4.0], requires_grad=True)
b = torch.tensor([1.0], requires_grad=True)

x = torch.tensor([2.0])
y_true = torch.tensor([7.0])

y_pred = w * x + b

loss = (y_pred - y_true) ** 2

loss.backward()

print("Gradient of w (w.grad): ", w.grad)
print("Gradient of b (b.grad): ", b.grad)
```
