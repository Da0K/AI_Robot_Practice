## 4. 함수 문제

```python
# 정육면체의 가로, 세로, 높이를 입력받아 부피를 구하는 함수 volume() 프로그램 만들기

def is_positive(n):
    if n > 0:
        return 1
    else:
        return 0

def volume(x, y, z):
    if is_positive(x) and is_positive(y) and is_positive(z):
        print( x * y * z )
    else:
        print("길이는 양수 값입니다.")

x, y, z = input("가로 세로 높이를 입력해주세요.").split()
volume(int(x), int(y), int(z))
```

```python
# 최솟값과 최댓값을 입력받아 그 사이 모든 정수의 총합을 구하는 함수 add_range() 프로그램 만들기

def is_positive(n):
    if (n > 0):
        return 1
    else:
        return 0

def add_range(min_num, max_num):
    if(is_positive(min_num) and is_positive(max_num)):
        sum = 0
        for i in range(min_num, max_num+1):
            sum += i
        print(sum)
    else:
        print("양수 값을 입력해주세요.")

min_num = int(input("최솟값을 입력하세요 : "))
max_num = int(input("최댓값을 입력하세요 : "))
add_range(min_num, max_num)
```

# 1. 권투 게임
두 명의 권투 선수의 능력치를 랜덤으로 설정하여 9라운드까지 대결하여 결과를 출력하는 권투 게임 프로그램 작성

* 각 선수의 능력치를 dictionary를 활용하여 정의
    - Hit_point : 350~500사이 정수로 무작위 생성
    - Punch_power : 7~10 사이의 정수로 무작위 생성
    - Effective_blow_rate : 3~7 사이의 정수로 무작위 생성
* 각 라운드 별 Hit_point의 감소 공식은 아래와 같음
    - Hit_point – 상대 선수의 (Punch_power * Effective_blow_rate)
* 승리 조건
    - KO 승(Knock-out): 9 라운드 이전에 한 선수의 Hit_point가 0 이하가 되면 상대 선수의 승리
    - 판정승(Win by decision): 9 라운드 종료 후 남은 Hit_point가 많은 선수가 승리
    - 무승부(Draw): 9 라운드 종료 후 두 선수의 Hit_point가 같은 경우나 라운드가 끝나기 전에 두 선수가 동시에 HP가 음수가 되는 경우
* 출력 정보
    - 초기 선수 능력치, 승리 선수 및 승리 방법(라운드), 두 선수의 남은 Hit_point

```python
import random

A = {
    'Hit_point' : random.randint(350, 500),
    'Punch_power' : random.randint(7, 10),
    'Effective_blow_rate' : random.randint(3, 7)
}

B = {
    'Hit_point' : random.randint(350, 500),
    'Punch_power' : random.randint(7, 10),
    'Effective_blow_rate' : random.randint(3, 7)
}

print(f'A의 능력치 : {A}')
print(f'B의 능력치 : {B}')
```

```python
A_Hit_point = A['Hit_point']
B_Hit_point = B['Hit_point']

i = 1
win_result = None

while(i <= 9):
    A['Hit_point'] -= B['Punch_power'] * B['Effective_blow_rate']
    B['Hit_point'] -= A['Punch_power'] * A['Effective_blow_rate']

    if A['Hit_point'] <= 0 and B['Hit_point'] <= 0:
        # 무승부(Draw)
        win_result = 'Draw'
        winner = 'None'
        break

    elif A['Hit_point'] <= 0:
        # B가 이긴 경우
        win_result = 'B win'
        winner = 'B'
        break

    elif B['Hit_point'] <= 0:
        # A가 이긴 경우
        win_result = 'A win'
        winner = 'A'
        break

if i==9 and win_result==None:
    # 판정승
    if A['Hit_point'] == B['Hit_point']:
        # 무승부
        win_result = 'Draw'
        winner = 'None'
    elif A['Hit_point'] < B['Hit_point']:
        # B가 이긴 경우
        win_result = 'B win'
        winner = 'B'
    else:
        # A가 이긴 경우
        win_result = 'A win'
        winner = 'A'

print(f"A의 초기 능력치 : Hit_point={A_Hit_point}, Punch_power={A['Punch_power']}, Effective_blow_rate={A['Effective_blow_rate']}")
print(f"B의 초기 능력치 : Hit_point={B_Hit_point}, Punch_power={B['Punch_power']}, Effective_blow_rate={B['Effective_blow_rate']}")
print(f'승부 결과={win_result}, 승자={winner}')
print(f"A 선수의 남은 Hit_point={A['Hit_point']}, B 선수의 남은 Hit_point={B['Hit_point']}")
```
