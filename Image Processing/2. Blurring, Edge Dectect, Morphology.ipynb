```python
import numpy as np
import cv2 as cv
```

```python
# Average Bluuring

img = cv.imread('images/green.png')
kernel = np.ones((5, 5), np.float32)/25 
dst = cv.filter2D(img, -1, kernel)

cv.imshow('Original', img)
cv.imshow('Result', dst)

cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# Average Bluuring

img = cv.imread('images/test.png')
img_blur = cv.blur(img, (11, 11))

cv.imshow('Original', img)
cv.imshow('Result', img_blur)

cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# Gaussian Bluuring

img = cv.imread('images/test.png')
img_blur = cv.GaussianBlur(img, (11, 11), 0)

cv.imshow('Original', img)
cv.imshow('Result', img_blur)

cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# Median Bluuring

img = cv.imread('images/median.png')
img_blur = cv.medianBlur(img, 5)

cv.imshow('Original', img)
cv.imshow('Result', img_blur)

cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# Bilateral Bluuring

img = cv.imread('images/texture.png')
img_blur = cv.bilateralFilter(img, 9, 75, 75)

cv.imshow('Original', img)
cv.imshow('Result', img_blur)

cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# Canny Edge Dectection

img_gray = cv.imread('images/house.png', cv2.IMREAD_GRAYSCALE)

cv.imshow('Original', img_gray)

img_canny = cv2.Canny(img_gray, 50, 150)
cv.imshow('Canny Edge without Blurring', img_canny)

img_gray = cv2.blur(img_gray, (3, 3))

img_canny = cv2.Canny(img_gray, 50, 150)
cv.imshow('Canny Edge with Blurring', img_canny)

cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# Erosion

img_gray = cv.imread('images/test3.png', cv2.IMREAD_GRAYSCALE)

# 다양한 크기의 kernel
kernel = cv.getStructuringElement(cv.MORPH_RECT, (1, 1))
img_result1x1 = cv.erode(img_gray, kernel, iterations=1)

kernel = cv.getStructuringElement(cv.MORPH_RECT, (3, 3))
img_result3x3 = cv.erode(img_gray, kernel, iterations=1)

kernel = cv.getStructuringElement(cv.MORPH_RECT, (5, 5))
img_result5x5 = cv.erode(img_gray, kernel, iterations=1)

# 반복 횟수 다르게

kernel = cv.getStructuringElement(cv.MORPH_RECT, (3, 3))
img_result3x3_iter2 = cv.erode(img_gray, kernel, iterations=2)

kernel = cv.getStructuringElement(cv.MORPH_RECT, (3, 3))
img_result3x3_iter3 = cv.erode(img_gray, kernel, iterations=3)

cv.imshow('Input', img_gray)
cv.imshow('Result 1x1', img_result1x1)
cv.imshow('Result 3x3', img_result3x3)
cv.imshow('Result 5x5', img_result5x5)

cv.imshow('Result 3x3 iter 2', img_result3x3_iter2)
cv.imshow('Result 3x3 iter 3', img_result3x3_iter3)

cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# Dilation

img_gray = cv.imread('images/test3.png', cv2.IMREAD_GRAYSCALE)

# 다양한 크기의 kernel
kernel = cv.getStructuringElement(cv.MORPH_RECT, (1, 1))
img_result1x1 = cv.dilate(img_gray, kernel, iterations=1)

kernel = cv.getStructuringElement(cv.MORPH_RECT, (3, 3))
img_result3x3 = cv.dilate(img_gray, kernel, iterations=1)

kernel = cv.getStructuringElement(cv.MORPH_RECT, (5, 5))
img_result5x5 = cv.dilate(img_gray, kernel, iterations=1)

# 반복 횟수 다르게

kernel = cv.getStructuringElement(cv.MORPH_RECT, (3, 3))
img_result3x3_iter2 = cv.dilate(img_gray, kernel, iterations=2)

kernel = cv.getStructuringElement(cv.MORPH_RECT, (3, 3))
img_result3x3_iter3 = cv.dilate(img_gray, kernel, iterations=3)

cv.imshow('Input', img_gray)
cv.imshow('Result 1x1', img_result1x1)
cv.imshow('Result 3x3', img_result3x3)
cv.imshow('Result 5x5', img_result5x5)

cv.imshow('Result 3x3 iter 2', img_result3x3_iter2)
cv.imshow('Result 3x3 iter 3', img_result3x3_iter3)

cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# Opening

img_gray = cv.imread('images/test4.png', cv2.IMREAD_GRAYSCALE)

kernel = cv.getStructuringElement(cv.MORPH_RECT, (3, 3))
img_result = cv.morphologyEx(img_gray, cv.MORPH_OPEN, kernel)

cv.imshow('Input', img_gray)
cv.imshow('Result', img_result)

cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# Closing

img_gray = cv.imread('images/test5.png', cv2.IMREAD_GRAYSCALE)

kernel = cv.getStructuringElement(cv.MORPH_RECT, (11, 11))
img_result = cv.morphologyEx(img_gray, cv.MORPH_CLOSE, kernel)

cv.imshow('Input', img_gray)
cv.imshow('Result', img_result)

cv.waitKey(0)
cv.destroyAllWindows()
```
