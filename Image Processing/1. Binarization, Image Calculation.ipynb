```python
import cv2 as cv
import sys

# Binary Threshold

img_color = cv.imread("images/grayscale.png", cv.IMREAD_COLOR)

if img_color is None:
    print("이미지 파일을 읽을 수 없습니다.")
    exit(1)

img_gray = cv.cvtColor(img_color, cv.COLOR_BGR2GRAY)

retval, img_binary = cv.threshold(img_gray, 127, 255, cv.THRESH_BINARY)

cv.imshow("Grayscale", img_gray)
cv.imshow('Binary', img_binary)

cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# Adacptive Threshold

img_color = cv.imread("images/copy.png", cv.IMREAD_COLOR)

if img_color is None:
    print("이미지 파일을 읽을 수 없습니다.")
    exit(1)

img_gray = cv.cvtColor(img_color, cv.COLOR_BGR2GRAY)

img_binary = cv.adaptiveThreshold(img_gray, 255, cv.ADAPTIVE_THRESH_MEAN_C, cv.THRESH_BINARY, 5, 4)

cv.imshow("Grayscale", img_gray)
cv.imshow('Binary', img_binary)

cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# Blending

alpha = 0.0
beta = 1.0

while alpha <= 1.0:
    img1 = cv.imread('images/beach.png', cv.IMREAD_COLOR)
    img2 = cv.imread('images/cat.png', cv.IMREAD_COLOR)

    dst = cv.addWeighted(img1, alpha, img2, beta, 0)

    print(alpha, " ", beta)

    cv.imshow('dst', dst)
    cv.waitKey(0)

    alpha = round(alpha + 0.1, 1)
    beta = round(beta - 0.1, 1)

cv.destroyAllWindows()
```

```python
# Subtract

img_background = cv.imread('images/background1.png', cv.IMREAD_GRAYSCALE)
img_object = cv.imread('images/object.png', cv.IMREAD_GRAYSCALE)

img_sub = cv.subtract(img_object, img_background)

retval, img_binary = cv.threshold(img_sub, 50, 255, cv.THRESH_BINARY)

cv.imshow("Background", img_background)
cv.imshow('Object', img_object)
cv.imshow("Sub", img_sub)
cv.imshow('binary', img_binary)

cv.waitKey(0)
cv.destroyAllWindows()
```
